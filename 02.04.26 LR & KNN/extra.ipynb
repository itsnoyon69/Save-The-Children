```python
import numpy as np
import pandas as pd
import math

from fontTools.subset import subset
from scipy.signal import impulse
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.linear_model import LogisticRegression
from sklearn.impute import SimpleImputer
from sklearn.metrics import accuracy_score
from sklearn.metrics import confusion_matrix
```

```python
dataset=pd.read_csv('LR/titanic.csv')
dataset.head()
```

```python
dataset=dataset[['name','sex','age','pclass','cabin','survived']]
dataset.shape
```

```python
dataset
```

```python
dataset.describe()
```

```python
print(dataset.isna().sum(axis=0))
```

```python
data=dataset.dropna(subset=['name','sex','pclass','survived'])
data
```

```python
print(data.isna().sum(axis=0))
```

```python
data["sex"] = data["sex"].map({"male":1,"female":0})

data["sex"]
```

```python
le = LabelEncoder()
data['name'] = le.fit_transform(data['name'])
data['cabin']=le.fit_transform(data['cabin'])
```

```python
data.head()
```

```python
feature=data[['name','sex','age','pclass','cabin']]
target=data['survived']
imputer=SimpleImputer(missing_values=np.nan, strategy='mean')
feature = imputer.fit_transform(feature)
```

```python
feature_train, feature_test, target_train, target_test = train_test_split(feature,target)
```

```python
model = LogisticRegression()
model.fit(feature_train, target_train)
predictions = model.predict(feature_test)
```

```python
print(confusion_matrix(target_test,predictions))
print(accuracy_score(target_test,predictions))
```
