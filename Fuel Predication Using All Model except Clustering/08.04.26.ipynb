Data Prerocessing

```python
le = LabelEncoder()

for col in ['country','region','income_level','subsidy_level']:
    data[col] = le.fit_transform(data[col])
data
```

Split Data

```python
#x = data.drop("petrol_usd_liter", axis=1)
x = data.drop(["petrol_usd_liter", "diesel_usd_liter","lpg_usd_liter"], axis=1)
#x = data['diesel_usd_liter']
y= data['petrol_usd_liter']
#x = x.values.reshape(-1,1)
print(x.shape)
print(y.shape)
```

```python
x_test,x_train,y_test,y_train=train_test_split(x,y,test_size=0.2,random_state=42)
```

Regression Model Train

```python
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.svm import SVR
from sklearn.neighbors import KNeighborsRegressor
```

```python
models = {
    "Linear Regression": LinearRegression(),
    "Decision Tree Regression": DecisionTreeRegressor(),
    "Random Forest Regression": RandomForestRegressor(),
    "Support Vector Regression": SVR(),
    "K Nearest Neighbors Regression": KNeighborsRegressor()
}
```

```python
for name, model in models.items():
    model.fit(x_train, y_train)
    y_pred = model.predict(x_test)

    r2 = r2_score(y_test, y_pred)
    mse = mean_squared_error(y_test, y_pred)
    rmse = np.sqrt(mse)
    mae = mean_absolute_error(y_test, y_pred)

    print("Model:", name)
    print("R2 Score:", r2)
    print("MSE:", mse)
    print("RMSE:", rmse)
    print("MAE:", mae)
    print("Train:", model.score(x_train, y_train))
    print("Test:", model.score(x_test, y_test))
    print('\n')
    print("------------------------------------------------------\n")
```

```python
x_new_test,x_new_train,y_new_test,y_new_train=train_test_split(x_new,y_new,test_size=0.2,random_state=42)
```

Model Training

```python
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import classification_report
```

```python
modelss = {
    "Logistic Regression": LogisticRegression(),
    "Decision Tree Classification": DecisionTreeClassifier(),
    "Random Forest Classification": RandomForestClassifier(n_estimators=200, max_depth=10),
    "Support Vector Classification": SVC(),
    "K Nearest Neighbors Classification": KNeighborsClassifier(),
    "Naive Bayes Classification": GaussianNB()
}
```

```python
for name, model in modelss.items():
    model.fit(x_new_train, y_new_train)
    y_new_pred = model.predict(x_new_test)
    ac = accuracy_score(y_new_test, y_new_pred)
    cm = confusion_matrix(y_new_test, y_new_pred)
    cr = classification_report(y_new_test, y_new_pred)

    print("Model:", name)
    print("Accuracy:", ac)
    print("Train:", model.score(x_new_train, y_new_train))
    print("Test:", model.score(x_new_test, y_new_test))
    print("\nConfusion Matrix:\n", cm)
    print("\nClassification Report:\n", cr)

    train_score = model.score(x_new_train, y_new_train)
    test_score = model.score(x_new_test, y_new_test)
        #overfitting checking
    if train_score - test_score > 0.1:
        print("Overfitting detected")
    else:
        print("No strong overfitting")

    #Shuffle Test
    y_new_test_shuffled = np.random.permutation(y_new_test)
    shuffled_score = model.score(x_new_test, y_new_test_shuffled)

    print("\nOriginal Score:", test_score)
    print("Shuffled Score:", shuffled_score)

    if abs(test_score - shuffled_score) < 0.05:
        print("Possible Data Leakage / Copying")
    else:
        print("Model is learning real patterns")

    #Cross Validation (BEST CHECK)
    cv_scores = cross_val_score(model, x_new_train, y_new_train, cv=5)

    print("\nCross Validation Scores:", cv_scores)
    print("CV Mean:", cv_scores.mean())

    if cv_scores.mean() < 0.7:
        print("Weak model")
    else:
        print("Stable model")

    print("\n" + "-"*60 + "\n")

```
